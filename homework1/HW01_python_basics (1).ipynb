1 Приведите все города к единому виду ("Almaty", "Astana", …) и сохраните в новый список cities_clean.

````python
cities_clean = [city.title() for city in cities]
print(cities_clean)
````

2 Преобразуйте amounts из строк в числа: список amounts_int.

````python
amounts_int = [int(amount) for amount in amounts]
print(amounts_int)
````

3 Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).

````python
print(len(amounts_int))
print(sum(amounts_int))
print(round(sum(amounts_int) / len(amounts_int)))
````

4 Сколько заказов не оплачено и на какую сумму?

````python
unpaid_amounts = [amount for amount, status in zip(amounts_int, paid)
                  if not status]
print(len(unpaid_amounts))
print(sum(unpaid_amounts))
````

6 Выручка (только оплаченные) по Алматы и по Астане отдельно.

````python
for city_name in ["Almaty", "Astana"]:
    revenue = sum(
        amount
        for city, amount, status in zip(cities_clean, amounts_int, paid)
        if city == city_name and status
    )
    print(city_name, revenue)
````

7 Номер (индекс) и город самого крупного оплаченного заказа.

````python
unpaid_index = None
for i in range(len(amounts_int)):
    if not paid[i]:
        if unpaid_index is None or amounts_int[i] > amounts_int[unpaid_index]:
            unpaid_index = i
print(cities_clean[unpaid_index], amounts_int[unpaid_index])
````

8 Доля заказов выходного дня (сб, вс) в процентах.

````python
weekend_count = sum(day in ("сб", "вс") for day in days)
print(weekend_count / len(days) * 100, "%")
````

9 Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью while посчитайте, через сколько месяцев он накопит.

````python
months = 0
savings = 0
while savings < 400000:
    savings += 15000
    months += 1
print(months)
````

9.1 А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
months = 0
savings = 0
monthly_payment = 15000
while savings < 400000:
    savings += monthly_payment
    monthly_payment += 1000
    months += 1
print(months)
````

10 Без словарей и библиотек постройте два списка unique_cities и revenue_by_city (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.

````python
unique_cities = []
revenue_by_city = []
````

````python
for city, amount, status in zip(cities_clean, amounts_int, paid):
    if city not in unique_cities:
        unique_cities.append(city)
        revenue_by_city.append(0)
````

````python
    if status:
        index = unique_cities.index(city)
        revenue_by_city[index] += amount
````

````python
for i in range(len(revenue_by_city)):
    for j in range(i + 1, len(revenue_by_city)):
        if revenue_by_city[j] > revenue_by_city[i]:
            revenue_by_city[i], revenue_by_city[j] = revenue_by_city[j], revenue_by_city[i]
            unique_cities[i], unique_cities[j] = unique_cities[j], unique_cities[i]
````

````python
print(unique_cities)
print(revenue_by_city)
````

11 Найдите медиану суммы заказа без statistics и numpy. Учтите чётное число элементов.

````python
sorted_amounts = sorted(amounts_int)
middle = len(sorted_amounts) // 2
median = (sorted_amounts[middle - 1] + sorted_amounts[middle]) / 2
print("Медиана:", median)
````

12 Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как █.
пн | ████  20490
вт | ...

````python
for day in ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]:
    revenue = sum(
        amount
        for order_day, amount, status in zip(days, amounts_int, paid)
        if order_day == day and status
    )
    print(f"{day} | {'█' * (revenue // 5000)} {revenue}")
````
