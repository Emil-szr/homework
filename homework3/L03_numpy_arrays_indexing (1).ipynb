## 1. Ссылки и копии
`b = a` не копирует список: это второе имя того же объекта. Копия — `a[:]` или `a.copy()`.

````python
a = [1, 2, 3]
b = a          # то же самое
c = a.copy()   # копия
b.append(4)
print(a, c)
````

````python
# ПРЕДСКАЖИТЕ: что напечатается?
grid = [[0] * 3] * 3      # три ссылки на ОДИН список
grid[0][0] = 1
print(grid)

grid_ok = [[0] * 3 for _ in range(3)]   # три разных списка
grid_ok[0][0] = 1
print(grid_ok)
````

````python
grid[1][0] = 2
print(grid)
````

## 4. Индексация и срезы

````python
arr = np.arange(0, 11)
print(arr)
print(arr[8])
print(arr[1:5], arr[:5], arr[5:])
print(arr[3:8:2], arr[::-1], arr[9:2:-2])
````

````python
# Присваивание в срез: одно значение записывается во все позиции
arr[0:5] = 100
print(arr)
arr = np.arange(0, 11)    # вернём как было
````

### Срез — не копия! (как `b = a` у списков)

````python
# ПРЕДСКАЖИТЕ: изменится ли arr?
arr = np.arange(0, 11)
print(arr)
slice_of_arr = arr[0:6]
print(slice_of_arr)
slice_of_arr[:] = 99
print(slice_of_arr)
print(arr)
````

````python
arr = np.arange(0, 11)
arr_copy = arr.copy()      # настоящая копия
arr_copy[:] = 0
print(arr, arr_copy)
````

### Двумерные массивы: `[строка, столбец]`
```
         столбец 0  1  2
строка 0       [ 5, 10, 15]
строка 1       [20, 25, 30]
строка 2       [35, 40, 45]
```

````python
arr_2d = np.array([[5, 10, 15], [20, 25, 30], [35, 40, 45]])
print(arr_2d[1])           # вторая строка
print(arr_2d[1, 0])        # строка 1, столбец 0 → 20 (то же, что arr_2d[1][0])
print(arr_2d[:2, 1:])      # первые 2 строки, столбцы с 1-го
print(arr_2d[:, 2])        # весь столбец 2
````

### Маски: выбор по условию

````python
arr = np.arange(1, 11)
arr
````

````python
arr > 4
````

````python
arr[arr > 4]
````

````python
(arr != 4)
````

````python
(arr < 5)
````

````python
(arr != 4) & (arr < 5)
````

````python
(arr == 4) | (arr > 8)
````

````python
arr = np.arange(1, 11)
print(arr > 4)                          # массив True/False
print(arr[arr > 4])                     # только подходящие
print(arr[(arr != 4) & (arr < 5)])      # И: &, скобки обязательны
print(arr[(arr == 4) | (arr > 8)])      # ИЛИ: |
print((arr > 4).sum())                  # сколько True: True считается как 1
print((arr > 4).mean())                 # доля True
````

````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````
