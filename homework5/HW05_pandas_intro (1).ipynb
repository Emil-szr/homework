## [Челлендж] 7. Выбросы по правилу IQR
Для оплаченных заказов без TEST: `IQR = Q3 − Q1`, граница = `Q3 + 1.5 * IQR`. Сколько заказов выше границы, какая это доля и какая у них категория чаще всего?
Сравните с ДЗ 4: по z-оценке выбросов было 87. Почему правила дают разное число?

````python
paid_no_test = (df["status"] == "оплачен") & (df["customer_id"] != "TEST")
````

````python
amounts = df.loc[paid_no_test, "amount"]
````

````python
q1 = np.percentile(amounts, 25)
````

````python
q3 = np.percentile(amounts, 75)
````

````python
iqr = q3 - q1
````

````python
limit = q3 + 1.5 * iqr
````

````python
outliers = df.loc[paid_no_test & (df["amount"] > limit)]
````

````python
limit, len(outliers), round(len(outliers) / len(amounts) * 100, 1)
````

````python
outliers["category"].value_counts()
````

````python
граница 37905 выше неё 320 заказов или 6.5%
Чаще всего встречается электроника  308 заказов
По правилу IQR выделено 320 дорогих заказов почти все они относятся к электронике
В дз 4 z оценка дала 87 потому что она сравнивает суммы со средним и стандартным отклонением
а IQR  с границей на основе 25 и 75 перцентилей
Еще здесь повторные строки заказов еще не удалены поэтому числа не полностью сопоставимы
````
