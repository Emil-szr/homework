## 1. Функции
Функция — именованный кусок кода, который можно вызывать много раз. Исправили один раз — работает везде.

- `def имя(параметры):` — объявление
- `return` — **возвращает** значение. `print` только **показывает** его, вернуть его нельзя.

```python
def to_num(s):
    """Строку вида '15 990' или '8070.0' превращает в число."""
    return float(s.replace(" ", ""))

print(to_num("15 990"))
print(to_num("8070.0"))
print(to_num("15 990") * 2)
#print(to_num("abc"))
```

```python
# return против print: найдите разницу
def show_double(x):
    print(x * 2)

def get_double(x):
    return x * 2

a = show_double(5)
b = get_double(5)
print("a =", a, "| b =", b)
```

```python
# параметр по умолчанию
def segment(amount, big=30000, mid=5000):
    if amount >= big:
        return "крупный"
    elif amount >= mid:
        return "средний"
    else:
        return "мелкий"

print(segment(32000), segment(7700), segment(870))
print(segment(32000, big=50000))   # меняем порог только для этого вызова
```

## 2. Словари: «по каждому X»
Словарь хранит пары **ключ → значение**. Ключи уникальны, поиск по ключу мгновенный.

```python
order = {"order_id": 101674, "city": "Алматы", "amount": 8070.0, "status": "оплачен"}
print(order["city"])
print(order.get("promo"))          # ключа нет → None, без ошибки
print(order.get("promo", "нет"))   # значение по умолчанию
order["promo"] = "AUTUMN"          # добавить или изменить
print(list(order.keys()))
for k, v in order.items():
    print(f"{k:>10}: {v}")
```

### Паттерн «словарь-накопитель» — главное, что нужно унести с занятия

```python
revenue = {}
for c, a, p in zip(cities, amounts, paid):
    if p:
        revenue[c] = revenue.get(c, 0) + a
print(revenue)

best = max(revenue, key=revenue.get)       # ключ с максимальным значением
print("Лидер:", best, revenue[best])

for c, v in sorted(revenue.items(), key=lambda t: t[1], reverse=True):
    print(f"{c:<10} {v:>7,} ₸")
```

```python
# setdefault: словарь списков — все суммы по каждому городу
by_city = {}
for c, a in zip(cities, amounts):
    by_city.setdefault(c, []).append(a)
    print(by_city)
print(by_city)
```

[Стандарт] Выручка оплаченных заказов по городам и город-лидер.

```python
revenue = {}

for city, amount, is_paid in zip(cities, amounts, paid):
    if is_paid:
        revenue[city] = revenue.get(city, 0) + amount

print(revenue)
print(max(revenue, key=revenue.get))
```

[Челлендж] Средний чек по каждому городу по всем заказам (через словарь списков). Почему у Астаны он больше 30 000 ₸, хотя оплачено там только 4 500 ₸?

```python
city_amounts = {}

for city, amount in zip(cities, amounts):
    city_amounts.setdefault(city, []).append(amount)

for city, values in city_amounts.items():
    average = sum(values) / len(values)
    print(city, average)
```

```python
def segment(amount, big=30000, mid=5000):
    if amount >= big:
        return "крупный"
    elif amount >= mid:
        return "средний"
    else:
        return "мелкий"
```

```python
paid_amounts = [a for a, p in zip(amounts, paid) if p]
print(paid_amounts)

print(amounts)
segments = [segment(a) for a in amounts]
print(segments)

upper = {c.upper() for c in cities}       # set comprehension: уникальные значения
print(upper)
```

```python
# Ловушка dict comprehension: ключи уникальны, поэтому побеждает ПОСЛЕДНЕЕ значение
last = {c: a for c, a in zip(cities, amounts)}
print(last)   # где 15 990 у Almaty?
```

> Comprehension хорош для **преобразования** и **фильтра**. Для **накопления** (суммы по ключу) используйте обычный цикл со словарём.

- **[Челлендж]** Напишите `clean_city(s)`: словарь-справочник `{"almaty": "Алматы", ...}` + `.strip().lower()`. Посчитайте выручку оплаченных по 6 чистым городам. Какой город выглядит подозрительно и почему?

```python
# Практика 3
```

```python
print("Всего заказов:", len(rows))

for status in set(row["status"] for row in rows):
    count = sum(row["status"] == status for row in rows)
    print(status, count)
```

```python
import numpy as np
amounts = [float(row["amount"].replace(" ", "")) for row in rows]
print("Пример суммы как числа:", amounts[0])
paid_amounts = [
    float(row["amount"].replace(" ", ""))
    for row in rows
    if row["status"] == "оплачен"
]
print("Выручка оплаченных заказов:", np.sum(paid_amounts))
print("Разных значений city:", len(set(row["city"] for row in rows)))
```

```python
city_names = {
    "almaty": "алматы",
    "astana": "астана",
    "shymkent": "шымкент",
}
revenue = {}
for row in rows:
    city = row["city"].strip().lower()
    city = city_names.get(city, city)
    if city not in revenue:
        revenue[city] = 0
    if row["status"] == "оплачен":
        amount = float(row["amount"].replace(" ", ""))
        revenue[city] += amount
for city, total in sorted(revenue.items()):
    print(city.title(), "|", int(total), "₸")
```
