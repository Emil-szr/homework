1. Напишите функции to_num(s) и clean_city(s) (6 городов, латиница и разный регистр → кириллица с заглавной).

````python
def to_num(s):
    return float(s.replace(" ", ""))
def clean_city(s):
    city = s.strip().lower()
    latin_names = {
        "almaty": "алматы",
        "astana": "астана",
        "shymkent": "шымкент"
    }
    return latin_names.get(city, city).title()
````

2. Сколько заказов в каждой категории (category)? Отсортируйте по убыванию.

````python
category_counts = {}
for row in rows:
    category = row["category"]
    category_counts[category] = category_counts.get(category, 0) + 1

for category, count in sorted(
    category_counts.items(), key=lambda item: item[1], reverse=True
):
    print(category, count)
````

3. Сколько заказов в каждой категории (category)? Отсортируйте по убыванию.

````python
category_revenue = {}
paid_counts = {}
for row in rows:
    if row["status"] == "оплачен":
        category = row["category"]
        category_revenue[category] = category_revenue.get(category, 0) + to_num(row["amount"])
        paid_counts[category] = paid_counts.get(category, 0) + 1
for category in category_revenue:
    average = category_revenue[category] / paid_counts[category]
    print(f"{category} {category_revenue[category]:.0f} {average:.2f}")
````

4. Сколько заказов в каждом месяце? Месяц — это order_date[:7].

````python
month_counts = {}
for row in rows:
    month = row["order_date"][:7]
    month_counts[month] = month_counts.get(month, 0) + 1
for month, count in sorted(month_counts.items()):
    print(month, count)
````

5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.

````python
monthly_revenue = {}
for row in rows:
    if row["status"] == "оплачен":
        month = row["order_date"][:7]
        monthly_revenue[month] = monthly_revenue.get(month, 0) + to_num(row["amount"])
for month, revenue in sorted(monthly_revenue.items()):
    print(month, f"{revenue:.0f} ₸")
change = (monthly_revenue["2026-10"] - monthly_revenue["2026-09"]) / monthly_revenue["2026-09"] * 100
print(f"Изменение: {change:.1f}%")
````

6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.

````python
all_orders = {}
cancelled_orders = {}
for row in rows:
    city = clean_city(row["city"])
    all_orders[city] = all_orders.get(city, 0) + 1
    if row["status"] == "отменён":
        cancelled_orders[city] = cancelled_orders.get(city, 0) + 1
for city in sorted(all_orders):
    share = cancelled_orders.get(city, 0) / all_orders[city] * 100
    print(f"{city}: {share:.1f}%")
````

7. Топ-5 клиентов (customer_id) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.

````python
customer_revenue = {}
for row in rows:
    if row["status"] == "оплачен":
        customer = row["customer_id"]
        customer_revenue[customer] = customer_revenue.get(customer, 0) + to_num(row["amount"])
top_5 = sorted(customer_revenue.items(), key=lambda item: item[1], reverse=True)[:5]
for customer, revenue in top_5:
    print(customer, f"{revenue:.0f}")
````

8. Сколько уникальных клиентов в каждом городе? Подсказка: словарь множеств.

````python
customers_by_city = {}
for row in rows:
    city = clean_city(row["city"])
    if city not in customers_by_city:
        customers_by_city[city] = set()
    customers_by_city[city].add(row["customer_id"])
for city in sorted(customers_by_city):
    print(city, len(customers_by_city[city]))
````

9. Найдите полные дубли строк: превратите каждую строку в tuple(r.values()) и проверяйте через множество. Сколько их и в каком месяце?

````python
seen = set()
unique_rows = []
duplicates_by_month = {}
for row in rows:
    record = tuple(row.values())
    if record in seen:
        month = row["order_date"][:7]
        duplicates_by_month[month] = duplicates_by_month.get(month, 0) + 1
    else:
        seen.add(record)
        unique_rows.append(row)
print(len(rows) - len(unique_rows))
print(duplicates_by_month)
````

10. Пересчитайте выручку по месяцам без дублей и без клиента из задачи 7. Как изменился процент из задачи 5?

````python
clean_rows = [row for row in unique_rows if row["customer_id"] != "TEST"]
revenue_by_month = {}
for row in clean_rows:
    if row["status"] == "оплачен":
        month = row["order_date"][:7]
        revenue_by_month[month] = revenue_by_month.get(month, 0) + to_num(row["amount"])
for month, revenue in sorted(revenue_by_month.items()):
    print(month, f"{revenue:.0f} ₸")
change = (
    revenue_by_month["2026-10"] - revenue_by_month["2026-09"]
) / revenue_by_month["2026-09"] * 100
print(f"{change:.1f}%")
````

11. У какого клиента больше всего заказов (после удаления дублей)? Сколько?

````python
orders_by_customer = {}
for row in unique_rows:
    customer = row["customer_id"]
    orders_by_customer[customer] = orders_by_customer.get(customer, 0) + 1
customer = max(orders_by_customer, key=orders_by_customer.get)
print(customer, orders_by_customer[customer])
````
